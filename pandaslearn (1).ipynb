```python
data={
    "players":["virat","rohit","sehwag","dhoni","virat","rohit","sehwag","dhoni"],
    "matches":[100,200,300,400,100,200,300,400],
    "runs":[154,200,123,257,309,198,171,120],
    "balls":[50,100,120,120,150,80,120,100]
}
ipl=pd.DataFrame(data)
ipl
```

```python
ipl["strike_rate"]=(ipl["runs"]/ipl["balls"])*100
ipl
```

```python
ipl=ipl.loc[(ipl["players"]=="virat")&(ipl["runs"]>150)]
ipl
```

It seems that the `ipl` DataFrame was filtered in the previous step, which is why the aggregation only shows data for 'virat'. To get information for all players, I will re-create the `ipl` DataFrame from the original `data` and then recalculate the strike rate.

```python
data={
    "players":["virat","rohit","sehwag","dhoni","virat","rohit","sehwag","dhoni"],
    "matches":[100,200,300,400,100,200,300,400],
    "runs":[154,200,123,257,309,198,171,120],
    "balls":[50,100,120,120,150,80,120,100]
}
ipl=pd.DataFrame(data)
ipl["strike_rate"]=(ipl["runs"]/ipl["balls"])*100
ipl
```

Now that the `ipl` DataFrame contains all players, I will re-run the aggregation to get the summary for all of them.

```python
summ=ipl.groupby("players").agg(
    total_runs=("runs","sum"),
    average_runs=("runs","mean"),
    matches_played=("matches","count"), # Changed to count as sum might not be appropriate for matches
    highest_runs=("runs","max"),
    lowest_runs=("runs","min"),
    total_balls=("balls","sum")
).reset_index()
summ
```

```python
players=["virat","rohit","sehwag","dhoni","virat","rohit","sehwag","dhoni"]
runs=[154,200,123,257,309,198,171,120]
total={}
highest={}
for i in range(len(players)):
  player=players[i]
  run=runs[i]
  if player in total:
    total[player]=total[player]+run
  else:
    total[player]=run
  if player in highest:
    if highest[player]<run:
      highest[player]=run
  else:
    highest[player]=run
for player in total:
  print(f"{player}| {total[player]}| {highest[player]}")
```
