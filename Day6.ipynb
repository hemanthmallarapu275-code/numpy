#Data Cleaning

```python
import pandas as pd
import numpy as np
from datetime import datetime
```

```python
data = {
    "CustomerName": ["Rahul Sharma", "priya patel", "AMIT KUMAR", "neha gupta", "Vikas Jain", "priya patel", "sneha reddy", "rahul sharma ", "Pooja Mehta", None],
    "OrderDate": ["2023-01-01", "2023-01-01", "2023-01-02", "2023-01-02", "2023-01-03", "2023-01-03", "2023-01-04", "2023-01-04", "2023-01-05", None],
    "ProductName": ["Laptop", "Mouse", "Keyboard", "Monitor", "Laptop", "Webcam", "Headphones", "Mouse", "Keyboard", "Monitor"],
    "Quantity": ['1','2','1', '1', '-3', '1', '2', '1', '1', '1'],
    "Price": [1200, 25, 75, 250, 1150, 45, 150, 20, 80, 260],
    "City": ["Mumbai", "Delhi", "Bangalore", "Hyderabad", "Chennai", "Delhi", "Kolkata", "Mumbai", "Pune", "Ahmedabad"],
    "DeliveryDate": ["2023-01-05", "2023-01-03", "2023-01-06", "2023-01-07", "2023-01-08", "2023-01-05", "2023-01-07", "2023-01-06", "2023-01-09", "2023-01-08"]
}
df = pd.DataFrame(data)
df
```

```python
df.info()
df.isna().sum()
```

```python
df.dtypes
```

```python
df.duplicated().sum() # it displays duplicate rows
```

```python
df["CustomerName"]=df["CustomerName"].str.strip()#this will remove leading and trailing spaces
df
```

```python
df["CustomerName"]=df["CustomerName"].str.strip()
df["CustomerName"]=df["CustomerName"].str.capitalize()#this convert the first letter into uppercase
df
```

```python
df.dropna(inplace=True)
df
```

```python
df["City"]=df["City"].replace({"Mumbai":"Goa","Chennai":"Madras"})
df
```

```python
def parseDate(datestr):
  if pd.isna(datestr):
    return pd.NaT
  datestr=str(datestr).strip()
  formats=['%Y-%m-%d','%Y/%m/%d','%d-%m-%Y','%Y-%d-%m','%d/%m/%Y']
  for fmt in formats:
    try:
      return pd.to_datetime(datestr,format=fmt)
    except:
      continue
  return pd.NaT
df
```

```python
df["OrderDate"]=df["OrderDate"].apply(parseDate)
df["DeliveryDate"]=df["DeliveryDate"].apply(parseDate)
df
```

```python
df["Quantity"]=df["Quantity"].astype("int64")
df.dtypes
```

```python
print("Now we are going to check for the negative values in Quantity column")
df[df["Quantity"]==2]
```

```python
df["Quantity"]=df["Quantity"].abs() # converting the negative value to positive
df
```
