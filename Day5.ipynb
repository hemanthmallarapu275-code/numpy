```python
import pandas as pd
import numpy as np
```

```python
df={
    "name":["Rahul","Joy","Kumar","Mahi","Jhanvi","Shubham","Priya","Anit"],
    "age":[25,np.nan,35,40,50,np.nan,43,45],
    "city":["New York","Los Angeles","Chicago","Houston","hongkong","delhi",np.nan,"kolkata"],
    "sal":[50000,60000,70000,80000,30000,32000,45000,870000],
    "Experience":[2,5,7,np.nan,15,1,8,np.nan],
    "Dept":["IT","HR","Finance",np.nan,"HR","Finance","IT","HR"]
    }
df=pd.DataFrame(df)
df
```

```python
df.isna()
df.isnull() #displays true where data is missing
```

```python
df.notna() #Shows false where the data is missing
df.notna().sum() #it shows not null values
```

```python
df.notna().mean()*100
```

```python
df[df.isna().any(axis=1)] #displays the rows which contain even one empty element
df[df.isna().all(axis=1)] #to check the complete empty row
```

```python
df[df["age"].isna()] #displays complete rows in which the specific col values are empty
```

```python
df.dropna() #this function drops the rows or col from the dataset
df.dropna(subset=["age"]) #this drops the rows which have missing values in specific column
df
```

```python
df={
    "name":["Rahul","Joy","Kumar","Mahi","Jhanvi","Shubham","Priya","Anit"],
    "age":[25,np.nan,35,40,50,np.nan,43,45],
    "city":["New York","Los Angeles","Chicago","Houston","hongkong","delhi",np.nan,"kolkata"],
    "sal":[50000,60000,70000,80000,30000,32000,45000,870000],
    "Experience":[2,5,7,np.nan,15,1,8,np.nan],
    "Dept":["IT","HR","Finance",np.nan,"HR","Finance","IT","HR"]
    }
df=pd.DataFrame(df)
df
```

```python
df.dropna(axis=1)#displays not null cols
df.dropna(axis=0)#displays not null rows
```

```python
df["Experience"].fillna(0)
df["city"].fillna("Unknown")
```

```python
df.fillna({"Experience":0,"city":"Unknown"})
```

###Filling the empty values statistically

```python
df["age"].fillna(df["age"].mean())
df["age"].fillna(df["age"].median())
```

```python
df["city"].fillna(df["city"].mode())
df["Dept"].fillna(df["Dept"].mode()[0])
```

```python
df.replace("Finance","Fina")
```
